```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_huggingface.embeddings import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_text_splitters import CharacterTextSplitter
```

```python
from langchain_core.documents import Document

# Initializing LangChain Document objects for marquee PSL players
doc1 = Document(
        page_content="Babar Azam is a world-class top-order batter and the captain of Peshawar Zalmi. He finished as the top run-scorer of the 2026 PSL season with 588 runs, leading his team to their second championship title.",
        metadata={
            "player_name": "Babar Azam",
            "team": "Peshawar Zalmi",
            "role": "Batter",
            "captain": True,
            "season_runs": 588
        }
    )

doc2 = Document(
        page_content="Shaheen Shah Afridi is an elite left-arm fast bowler who captains the Lahore Qalandars. Known for his lethal swing and early breakthroughs in the powerplay overs.",
        metadata={
            "player_name": "Shaheen Shah Afridi",
            "team": "Lahore Qalandars",
            "role": "Bowler",
            "captain": True,
            "bowling_style": "Left-arm fast"
        }
    )

doc3 = Document(
        page_content="Mohammad Rizwan is a highly dependable wicketkeeper-batter who captains the Rawalpindiz franchise. He provides exceptional consistency at the top of the order and sharp leadership behind the stumps.",
        metadata={
            "player_name": "Mohammad Rizwan",
            "team": "Rawalpindiz",
            "role": "Wicketkeeper-Batter",
            "captain": True
        }
    )

doc4 = Document(
        page_content="Shadab Khan is a premier leg-spinning all-rounder and the long-standing captain of Islamabad United. He provides dynamic lower-order hitting and elite middle-overs control.",
        metadata={
            "player_name": "Shadab Khan",
            "team": "Islamabad United",
            "role": "All-Rounder",
            "captain": True
        }
    )

doc5 = Document(
        page_content="David Warner is an explosive international opening batsman who captained the Karachi Kings for the 2026 season, bringing massive powerplay scoring and veteran leadership.",
        metadata={
            "player_name": "David Warner",
            "team": "Karachi Kings",
            "role": "Batter",
            "captain": True
        }
    )

doc6 = Document(
        page_content="Sufiyan Muqeem is a rising left-arm wrist spinner playing for Peshawar Zalmi. He won the Player of the Series award in the 2026 tournament after finishing as the leading wicket-taker with 22 wickets.",
        metadata={
            "player_name": "Sufiyan Muqeem",
            "team": "Peshawar Zalmi",
            "role": "Bowler",
            "captain": False,
            "season_wickets": 22
        }
    )

# # Example: Inspecting a single LangChain Document structure
# print(f"Content: {psl_player_documents[0].page_content}")
# print(f"Metadata: {psl_player_documents[0].metadata}")
```

```python
docs = [doc1, doc2, doc3, doc4, doc5, doc6]
```

```python
vector_store = Chroma(
    embedding_function=HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2"),
    persist_directory="./chroma_db",
    collection_name="psl_players"
)
```

```python
# add documents
vector_store.add_documents(docs)
```

```python
# view documents
vector_store.get(include=['embeddings','documents','metadatas'])
```

```python
# Search documents
vector_store.similarity_search(
    query='Who among these are a bowler?',
    k=2
)
```

```python
# Search with Similarity Score
vector_store.similarity_search_with_score(
    query='Who among these are a bowler?',
    k=2
)
```

```python
# Meta Data filtering
vector_store.similarity_search(
    query='',
    filter={'role': 'Bowler'}
)
```

```python
from langchain_core.documents import Document

# Update Documents
update_doc1 = Document(
    page_content="Once again Babar Azam gets the captaincy of the Test cricket team because he is the first-class captain and batter.",
    metadata={
        "player_name": "Babar Azam",
        "team": "Pakistan",
        "role": "Batter",
        "captain": True
    },
    id='880d850a-901a-4607-94cc-11479ab3d6bd' # Add the document ID here
)

vector_store.update_documents(
    documents=[update_doc1], # Pass a list of Document objects
    ids=[update_doc1.id] # Pass a list of corresponding IDs
)
```

```python
# view documents
vector_store.get(include=['embeddings','documents','metadatas'])
```

```python
# delete documents
vector_store.delete(ids=['880d850a-901a-4607-94cc-11479ab3d6bd'])
```
